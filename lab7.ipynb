```python
import numpy as np

study_hours = np.array([2, 3, 4, 5, 6, 7, 8, 9])
marks = np.array([45, 50, 54, 60, 65, 70, 74, 80])

x_mean = np.mean(study_hours)
y_mean = np.mean(marks)

x_diff = study_hours - x_mean
y_diff = marks - y_mean

numerator = np.sum(x_diff * y_diff)
denominator = np.sqrt(np.sum(x_diff ** 2) * np.sum(y_diff ** 2))

correlation = numerator / denominator

print(correlation)
```

```python
import numpy as np
from scipy.stats import f_oneway

uni_selec = np.array([82, 84, 81, 83, 85])
chi_selec = np.array([85, 87, 86, 84, 88])
mut_selec = np.array([88, 89, 91, 90, 92])

uni_mean = np.mean(uni_selec)
chi_mean = np.mean(chi_selec)
mut_mean = np.mean(mut_selec)

print("Univerient Mean:", uni_mean)
print("Chi-Squre Mean:", chi_mean)
print("Mutual Mean:", mut_mean)

F_stat, p_value = f_oneway(uni_selec, chi_selec, mut_selec)

print("F-statistic:", F_stat)
print("p-value:", p_value)
```
