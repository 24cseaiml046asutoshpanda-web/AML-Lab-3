```python
# implementing the one R for providing the best attribute of the feature


import pandas as pd

df = pd.DataFrame({
    "rainfall": ["high", "high", "high", "low", "low", "low"],
    "soiltype": ["loary", "sandy", "clay", "loary", "sandy", "clay"],
    "yield": ["good", "good", "good", "poor", "poor", "good"]
})

print(df)
```

```python
import pandas as pd

def one_r_feature_selection(X, y):
    results = []
    all_rules = {}

    for feature in X.columns:
        rules = {}
        total_errors = 0

        # Process each distinct value of this feature
        for value in X[feature].dropna().unique():
            rows = X[feature] == value
            target_values = y[rows]

            # Count each target class
            class_counts = target_values.value_counts()

            # Select the majority target class
            predicted_class = class_counts.idxmax()

            # Count incorrect predictions
            errors = (target_values != predicted_class).sum()
            total_errors += errors

            rules[value] = {
                "prediction": predicted_class,
                "counts": class_counts.to_dict(),
                "errors": int(errors)
            }

        error_rate = total_errors / len(y)
        accuracy = 1 - error_rate

        results.append({
            "feature": feature,
            "errors": total_errors,
            "error_rate": error_rate,
            "accuracy": accuracy
        })

        all_rules[feature] = rules

    scores = (
        pd.DataFrame(results)
        .sort_values("error_rate")
        .reset_index(drop=True)
    )

    best_feature = scores.loc[0, "feature"]

    return best_feature, scores, all_rules
```

```python
X = df[["rainfall", "soiltype"]]
y = df["yield"]

best_feature, scores, rules = one_r_feature_selection(X, y)

print("Scores:")
print(scores)

print("\nBest feature:")
print(best_feature)

print("\nRules:")
for value, rule in rules[best_feature].items():
    print(
        f"IF {best_feature} = {value} "
        f"THEN yield = {rule['prediction']}"
    )
```
