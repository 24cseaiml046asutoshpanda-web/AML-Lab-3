```python
import numpy as np

study_hours = np.array([2, 3, 4, 5, 6, 7, 8, 9])
marks = np.array([45, 50, 54, 60, 65, 70, 74, 80])

x_mean = np.mean(study_hours)
y_mean = np.mean(marks)

x_diff = study_hours - x_mean
y_diff = marks - y_mean

numerator = np.sum(x_diff * y_diff)
denominator = np.sqrt(np.sum(x_diff ** 2) * np.sum(y_diff ** 2))

correlation = numerator / denominator

print(correlation)
```
