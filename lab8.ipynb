```python
import numpy as np

# Data
X = np.array([
    [1, 2, 5],
    [2, 3, 4],
    [3, 4, 3],
    [7, 6, 3],
    [8, 7, 2],
    [9, 8, 1]
])

y = np.array([0, 0, 0, 1, 1, 1])

C = np.corrcoef(np.column_stack((X, y)), rowvar=False)

r_y = np.abs(C[:3, 3])

pairs = [(0, 1), (0, 2), (1, 2)]

M = np.array([
    2 * np.mean(r_y[list(pair)]) /
    np.sqrt(2 + 2 * abs(C[pair]))
    for pair in pairs
])

print("M values:", M)

i = np.argmax(M)
print(f"M{i + 1} selected: x{pairs[i][0] + 1} and x{pairs[i][1] + 1}")
```

```python
import numpy as np

x1 = np.array([1,2,3,7,8,9])
x2 = np.array([2,3,4,6,7,8])
x3 = np.array([5,4,3,3,2,1])
y = np.array([0,0,0,1,1,1])

x1_mean = x1.mean()
x2_mean = x2.mean()
x3_mean = x3.mean()
y_mean = y.mean()

x1_c = x1 - x1_mean
x2_c = x2 - x2_mean
x3_c = x3 - x3_mean
y_c = y - y_mean


def correlation(x, y):
    numerator = np.sum(x * y)
    denominator = np.sqrt(np.sum(x**2) * np.sum(y**2))
    return numerator / denominator


r_x1_y = correlation(x1_c, y_c)
r_x2_y = correlation(x2_c, y_c)
r_x3_y = correlation(x3_c, y_c)

r_x1_x2 = correlation(x1_c, x2_c)
r_x1_x3 = correlation(x1_c, x3_c)
r_x2_x3 = correlation(x2_c, x3_c)

r_c_f1_mean = (r_x1_y + r_x2_y) / 2
r_c_f2_mean = (r_x1_y + r_x3_y) / 2
r_c_f3_mean = (r_x2_y + r_x3_y) / 2

M1 = (2 * r_c_f1_mean) / (2 + 2 * r_x1_x2) ** 0.5
M2 = (2 * r_c_f2_mean) / (2 + 2 * r_x1_x3) ** 0.5
M3 = (2 * r_c_f3_mean) / (2 + 2 * r_x2_x3) ** 0.5


if M1 > M2 and M1 > M3 :
  print("M1 is selected so x1 and x2 is selected")
elif M2 > M1 and M2 > M3 :
  print("M2 is selected so x1 and x3 is selected")
else :
  print("M3 is selected so x2 and x3 is selected")
```
